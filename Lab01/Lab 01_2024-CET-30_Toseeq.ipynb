07: Function: grab the domain from "user@domain.com".

```python
def get_domain(email):
    return email.split('@')[1]
print("Q7:",get_domain("user@domain.com"))
```

08: Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    return 'dog' in text.lower()
print("Q8:",find_dog("Is there a DOG here?")) 
```

Task 09:

09: Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(text):
    return text.lower().count("dog")
print("09:", count_dog("dog dog DOG"))
```

Task 10:

10: caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed,is_birthday):
    if is_birthday:
        speed = speed - 5
    if speed<=60:
        return "No Ticket"
    elif speed<=80:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10", caught_speeding (60,False))
print("Q10", caught_speeding(75,True))
print("Q10", caught_speeding(85,False))
```
