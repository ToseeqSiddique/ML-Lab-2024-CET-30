C:Seed 42: Roll 10,000 Ludo Dice

```python
np.random.seed(42)

dice = np.random.randint(1, 7, size=(10000, 2))

print("T3C:",dice)
```

D: Count Each Face Using np.unique.

```python
all_dice = dice.flatten()

faces, counts = np.unique(all_dice, return_counts=True)

print("T3D:","Dice Faces:", faces)
print("Counts:", counts)
```

E: Estimate P(Six) and Count Double Six.

```python
six_count = np.sum(all_dice == 6)

probability = six_count / len(all_dice)

double_six = np.sum((dice[:, 0] == 6) & (dice[:, 1] == 6))

print("T3E:","Probability of Six:", probability)
print("Double Six:", double_six)
```

H: Wettest Monsoon City and Monsoon Share %

```python
import numpy as np

data = np.array([
    [22, 32, 32, 20, 18, 74, 189, 177, 89, 10, 7, 7],
    [11, 6, 3, 11, 9, 25, 86, 105, 44, 13, 1, 6],
    [55, 93, 95, 58, 40, 78, 311, 317, 135, 34, 18, 26],
    [40, 70, 78, 68, 27, 25, 69, 69, 35, 27, 17, 17],
    [55, 58, 50, 40, 18, 3, 9, 5, 4, 5, 19, 35]
])

# Monsoon rainfall (July to September)
monsoon = data[:, 6:9]

# Total monsoon rainfall of each city
monsoon_total = monsoon.sum(axis=1)

# Wettest monsoon city
wettest_city = np.argmax(monsoon_total)

print("T7H: Monsoon Rainfall:", monsoon_total)
print("Wettest City Index:", wettest_city)

# Monsoon share percentage
monsoon_share = (monsoon.sum() / data.sum()) * 100

print("Monsoon Share Percentage:", monsoon_share)
```

Task 9: +5 capped at 100 with np.clip; pass %; who passed only because of grace; z-scores.

A: Add 5 grace marks.

```python
import numpy as np

marks = np.array ([45,38,32,48,55])

new_marks= marks + 5
print("T9A:",new_marks)
```

B: Cap at 100.

```python
new_marks = np.clip(new_marks, 0, 100)

print("T9B:",new_marks)
```

C: pass percantage:

```python
passed = np.sum(new_marks >= 40)

pass_percentage = (passed / len(new_marks)) * 100

print("T9C:",pass_percentage)
```

D: z_scores:

```python
z_scores = (new_marks - np.mean(new_marks)) / np.std(new_marks)

print("T9D:",z_scores)
```
